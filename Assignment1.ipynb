# **3) Clean & Normalize Skills (text processing)**

Make a function called clean_skills() that accepts skills in string format.  Converts text to lowercase
 Replaces commas with spaces  Removes extra spaces (normalize whitespace)  Optionally removes punctuation (you may use basic str.replace() or re module)
Apply clean_skills() to every entry in job_data['Skills'] and overwrite the list with cleaned
versions.

```python
def clean_skills(skills):
    # Convert text to lowercase
    skills = skills.lower()

    # Replace commas with spaces
    skills = skills.replace(",", " ")

    # Remove extra spaces
    skills = " ".join(skills.split())

    return skills


# Apply clean_skills() to every entry in Skills
job_data["Skills"] = [clean_skills(skill) for skill in job_data["Skills"]]

print(job_data["Skills"])

```

# **4) Skill Extraction & Frequency**
From the cleaned skills list, split each skill string into individual skill tokens.  Build a consolidated list of all skills that contains every skill token across all postings.

```python
# Create an empty list to store all individual skills
all_skills = []

# Go through each skill string
for skills in job_data["Skills"]:
    # Split the skills into individual skills
    skill_list = skills.split()

    # Add them to the consolidated list
    all_skills.extend(skill_list)


# Create a dictionary to store skill frequency
skill_count = {}

# Count each skill
for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1


# Sort skills by frequency in descending order
sorted_skills = sorted(
    skill_count.items(),
    key=lambda x: x[1],
    reverse=True
)


# Print top 5 skills
print("Top skills by frequency:")

for skill, count in sorted_skills[:5]:
    print(f"{skill}: {count}")
```

# **6) Skill Search Function (functions & conditional logic)**
Create a function, find_jobs_by_skill(skill_query), that returns a list of (Job_ID, Role, Company) for postings containing the skill_query token. ● The function must normalize the skill_query (lowercase, strip spaces)
● Use if/elif/else where appropriate

```python
def find_jobs_by_skill(skill_query):

    # Normalize the search query
    skill_query = skill_query.lower().strip()

    results = []

    # Check whether the query is empty
    if skill_query == "":
        return "Please enter a skill."

    # Go through each job
    for i in range(len(job_data["Skills"])):

        skills = job_data["Skills"][i]

        # Check if the skill exists
        if skill_query in skills.split():
            results.append((
                job_data["Job_ID"][i],
                job_data["Role"][i],
                job_data["Company"][i]
            ))

        elif skill_query == "python":
            # Python-specific check
            if "python" in skills.split():
                results.append((
                    job_data["Job_ID"][i],
                    job_data["Role"][i],
                    job_data["Company"][i]
                ))

        else:
            continue

    return results
```

### **7) Lambda Filtering & List Comprehension**

Use a lambda function and filter() (or a list comprehension with a lambda-like expression) to
extract job roles where the skill count (from skill_count) for any skill >= 2 — that is, roles
that include at least one relatively common skill. Example Output

```python
roles_common_skill = list(filter(
    lambda i: any(
        skill_count.get(skill, 0) >= 2
        for skill in job_data["Skills"][i].split()
    ),
    range(len(job_data["Skills"]))
))

roles_common_skill = [
    job_data["Role"][i]
    for i in roles_common_skill
]

print("Roles with at least one common skill:", roles_common_skill)
```

# **9) Save Cleaned Data to File (file handling & exceptions)**
Save the cleaned & structured job records to a file named job_skills.csv. Each line should contain:
Job_ID,Role, Company, Skills.Use try/except/else/finally to handle file I/O errors. Also, create a secondary file, top_skills.txt that lists top skills and counts.

```python
import csv

# Check the number of records
print("Job_ID:", len(job_data["Job_ID"]))
print("Role:", len(job_data["Role"]))
print("Company:", len(job_data["Company"]))
print("Skills:", len(job_data["Skills"]))



# 1. Save job data to CSV file


try:
    # Check that all lists have the same length
    if not (
        len(job_data["Job_ID"])
        == len(job_data["Role"])
        == len(job_data["Company"])
        == len(job_data["Skills"])
    ):
        raise ValueError("All job_data lists must have the same length.")

    file = open("job_skills.csv", "w", newline="")

    writer = csv.writer(file)

    # Write header
    writer.writerow(["Job_ID", "Role", "Company", "Skills"])

    # Write job records
    for job_id, role, company, skills in zip(
        job_data["Job_ID"],
        job_data["Role"],
        job_data["Company"],
        job_data["Skills"]
    ):
        writer.writerow([job_id, role, company, skills])

except Exception as e:
    print("Error while writing job_skills.csv:", e)

else:
    print("job_skills.csv created successfully.")

finally:
    try:
        file.close()
    except:
        pass

# 2. Save top skills to TXT file


try:
    file = open("top_skills.txt", "w")

    file.write("Top skills by frequency:\n")

    # Sort skills by frequency
    sorted_skills = sorted(
        skill_count.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Write top 5 skills
    for skill, count in sorted_skills[:5]:
        file.write(f"{skill}: {count}\n")

except Exception as e:
    print("Error while writing top_skills.txt:", e)

else:
    print("top_skills.txt created successfully.")

finally:
    try:
        file.close()
    except:
        pass
```
